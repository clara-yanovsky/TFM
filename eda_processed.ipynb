```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("processed_matches.csv")
```

```python
df.head()
df.shape
df.info()
df.isnull().sum().sum()
df.duplicated().sum()
```

```python
main_vars = [
    "home_rank_position",
    "away_rank_position",
    "home_rank_points",
    "away_rank_points",
    "rank_points_diff",
    "home_score",
    "away_score",
    "ranking_diff",
    "goal_diff",
    "result",
    "neutral",
    "has_shootout",
    "year"
]

df[main_vars].head()
```

```python
df[main_vars].describe(include="all")
```

```python
plt.figure(figsize=(10,4))
plt.hist(df["home_score"], bins=20)
plt.title("Distribución de goles del equipo local")
plt.xlabel("Goles")
plt.ylabel("Frecuencia")
plt.show()

plt.figure(figsize=(10,4))
plt.hist(df["away_score"], bins=20)
plt.title("Distribución de goles del equipo visitante")
plt.xlabel("Goles")
plt.ylabel("Frecuencia")
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.hist(df["home_rank_position"], bins=30)
plt.title("Distribución del ranking del equipo local")
plt.xlabel("Posición ranking")
plt.ylabel("Frecuencia")
plt.show()

plt.figure(figsize=(10,4))
plt.hist(df["away_rank_position"], bins=30)
plt.title("Distribución del ranking del equipo visitante")
plt.xlabel("Posición ranking")
plt.ylabel("Frecuencia")
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.hist(df["ranking_diff"], bins=30)
plt.title("Distribución de la diferencia de ranking")
plt.xlabel("ranking_diff")
plt.ylabel("Frecuencia")
plt.show()

plt.figure(figsize=(10,4))
plt.hist(df["goal_diff"], bins=20)
plt.title("Distribución de la diferencia de goles")
plt.xlabel("goal_diff")
plt.ylabel("Frecuencia")
plt.show()
```

```python
df["result"].value_counts()
```

```python
plt.figure(figsize=(8,4))
df["result"].value_counts().plot(kind="bar")
plt.title("Distribución de resultados")
plt.xlabel("Resultado")
plt.ylabel("Frecuencia")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="result", y="ranking_diff", data=df)
plt.title("Diferencia de ranking según resultado")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="result", y="goal_diff", data=df)
plt.title("Diferencia de goles según resultado")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="result", y="rank_points_diff", data=df)
plt.title("Diferencia de puntos FIFA según resultado")
plt.show()
```

```python
plt.figure(figsize=(10,4))
df["year"].hist(bins=30)
plt.title("Distribución de partidos por año")
plt.xlabel("Año")
plt.ylabel("Frecuencia")
plt.show()
```

```python
df["neutral"].value_counts()

plt.figure(figsize=(6,4))
df["neutral"].value_counts().plot(kind="bar")
plt.title("Partidos en campo neutral")
plt.xlabel("Neutral")
plt.ylabel("Frecuencia")
plt.show()
```

```python
df["has_shootout"].value_counts()

plt.figure(figsize=(6,4))
df["has_shootout"].value_counts().plot(kind="bar")
plt.title("Partidos con definición por shootout")
plt.xlabel("Has shootout")
plt.ylabel("Frecuencia")
plt.show()
```

```python
corr_vars = [
    "home_rank_position",
    "away_rank_position",
    "home_rank_points",
    "away_rank_points",
    "rank_points_diff",
    "home_score",
    "away_score",
    "ranking_diff",
    "goal_diff",
    "year",
    "neutral",
    "has_shootout",
    "ranking_scaled"
]

plt.figure(figsize=(12,8))
sns.heatmap(df[corr_vars].corr(), annot=True, cmap="coolwarm")
plt.title("Matriz de correlación de variables clave")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df["home_score"])
plt.title("Boxplot goles local")
plt.show()

plt.figure(figsize=(8,4))
sns.boxplot(x=df["away_score"])
plt.title("Boxplot goles visitante")
plt.show()

plt.figure(figsize=(8,4))
sns.boxplot(x=df["ranking_diff"])
plt.title("Boxplot diferencia de ranking")
plt.show()
```
