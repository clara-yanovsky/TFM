```python
# =========================================================
# 1. IMPORTAR LIBRERÍAS
# =========================================================
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# =========================================================
# 2. CARGAR DATASET ORIGINAL
# =========================================================
df = pd.read_csv("matches_processed_for_regression.csv")

# =========================================================
# 3. REVISIÓN INICIAL DEL DATASET
# =========================================================
print("=== PRIMERAS FILAS ===")
print(df.head())

print("\n=== DIMENSIONES DEL DATASET ===")
print(df.shape)

print("\n=== ESTRUCTURA DEL DATASET ===")
print(df.info())

print("\n=== NOMBRES DE COLUMNAS ===")
print(df.columns.tolist())

print("\n=== NULOS POR COLUMNA ===")
print(df.isnull().sum())

print("\n=== NULOS TOTALES ===")
print(df.isnull().sum().sum())

print("\n=== DUPLICADOS ===")
print(df.duplicated().sum())

print("\n=== ESTADÍSTICA DESCRIPTIVA ===")
print(df.describe(include="all"))

# =========================================================
# 4. LIMPIEZA BÁSICA
# =========================================================
# Eliminar duplicados si existieran
df = df.drop_duplicates()

# Si quieres eliminar nulos en caso de que existan, descomenta esta línea:
# df = df.dropna()

# =========================================================
# 5. VALIDAR NOMBRES DE COLUMNAS NECESARIAS
# =========================================================
required_cols = ["home_rank_position", "away_rank_position", "home_score", "away_score"]

missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    raise ValueError(f"Faltan estas columnas necesarias en el dataset: {missing_cols}")

# =========================================================
# 6. CREAR VARIABLES DERIVADAS
# =========================================================

# 6.1 Diferencia de ranking
# Nota: valores menores en ranking significan mejor posición
df["ranking_diff"] = df["home_rank_position"] - df["away_rank_position"]

# 6.2 Diferencia de goles
df["goal_diff"] = df["home_score"] - df["away_score"]

# 6.3 Variable objetivo: resultado del partido
df["result"] = np.where(
    df["goal_diff"] > 0, "win",
    np.where(df["goal_diff"] < 0, "lose", "draw")
)

# 6.4 Variable dummy de localía
df["is_home"] = 1

# =========================================================
# 7. ESCALAR RANKING
# =========================================================
# Se escala la variable home_rank_position para crear ranking_scaled
scaler = StandardScaler()
df["ranking_scaled"] = scaler.fit_transform(df[["home_rank_position"]])

# =========================================================
# 8. VALIDACIÓN FINAL DEL DATASET YA MODIFICADO
# =========================================================
print("\n=== NUEVAS COLUMNAS AGREGADAS ===")
new_cols = ["ranking_diff", "goal_diff", "result", "is_home", "ranking_scaled"]
print(df[new_cols].head())

print("\n=== DIMENSIONES FINALES DEL DATASET ===")
print(df.shape)

print("\n=== NULOS TOTALES FINALES ===")
print(df.isnull().sum().sum())

print("\n=== DUPLICADOS FINALES ===")
print(df.duplicated().sum())

print("\n=== TIPOS DE DATOS FINALES ===")
print(df.dtypes.tail(10))

# =========================================================
# 9. GUARDAR EL NUEVO CSV
# =========================================================
df.to_csv("processed_matches.csv", index=False)

print("\nArchivo guardado correctamente como: processed_matches.csv")
```
